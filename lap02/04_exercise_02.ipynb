(d) Tính phương sai của dãy số $\sigma^2 = \frac{\sum{\left(a_i - \mu\right)^2}}{N - 1}$ với $\mu = \frac{\sum{a_i}}{N}$ và độ lệch chuẩn $var = \sqrt{\sigma^2}$

(a) Sinh ngẫu nhiên dãy số có 100 phần tử theo phân bố đều với giá trị trong đoạn [1, 10] và vẽ đồ thị tần số xuất hiện <br/>

 $$
 y = f\left(x| \mu, \sigma\right) = \frac{1}{\sigma\sqrt{2\pi}}exp\left(-\frac{\left(x - \mu\right)^2}{2\sigma^2}\right)

 nơi mà $x \in [\mu - 5\sigma, \mu + 5\sigma]$ <br/>
 (c) Sinh ngẫu nhiên dãy số có 100000 phần tử theo phân bố chuẩn $N(\mu = 1, \sigma^2 = 1.0)$ và vẽ đồ thị tần số xuất hiện <br/> <br/>

```python
import matplotlib.pyplot as plt  # Khai báo thư viện pyplot trong matplotlib

""" CÁC BẠN LÀM BÀI Ở ĐÂY """
# Sử dụng hàm np.random.randint để sinh dãy số ngẫu nhiên
# Sử dụng hàm np.unique để trả về tần số xuất hiện
# Sinh 100 phần tử có giá trị [1, 10]
data = np.random.randint(low=1, high=11, size=100)
value, cnt = np.unique(data, return_counts=True)

print("a. Sinh ngẫu nhiên theo phân bố đều")
print(f"+ Gia tri: {value}")
print(f"+ So lan xuat hien: {cnt}")
print("+ Do thi ham so xuat hien")
plt.figure(figsize=(6,6))      #  kich thuoc ban ve
plt.bar(value, cnt)            #  ve cac khoi chu nhat theo day x, y
plt.xlim(0, 11)                #  gioi han truc x
plt.ylim(0, np.max(cnt) + 10)  #  gioi han truc y
plt.savefig("4a.png")          #  luu do thi
plt.show()                     #  hien thi do thi
```

```python
""" CÁC BẠN LÀM BÀI Ở ĐÂY """
# Sử dụm hàm np.random.rand() sinh số ngẫu nhiên trong [0,1) --> [0,1) * (b - a) + a --> [a, b)
# Sử dụng hàm sau để tính hàm xác suất chuẩn N(m, s): np.sqrt, np.exp

# Khởi tạo các tham số phân phối chuẩn
mu = 1.0
sigma = 1.0
x_min = mu - 5 * sigma
x_max = mu + 5 * sigma
# Sinh ngẫu nhiên N điểm x trong khoảng [x_min, x_max]
x = np.random.rand(100000) * (x_max - x_min) + x_min
# Tính giá trị hàm phân phối chuẩn tại mỗi điểm x
y = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-((x - mu) ** 2) / (2 * sigma ** 2))
print("b. Sinh cặp (x, y) với y làm hàm phân phối chuẩn")
print(f"+ x: {x_min} <= min [{np.min(x): .2f}] <= max [{np.max(x): .2f}] <= {x_max}")
print(f"+ Ve ham so y(x)")
plt.figure(figsize=(6,6))      # kich thuoc ban ve
plt.scatter(x, y)              # ve cac diem (xi, yi)
plt.savefig("4b.png")          #  luu do thi
plt.show()                     #  hien thi do thi
```

```python
""" CÁC BẠN LÀM BÀI Ở ĐÂY """
# Sử dụng hàm np.random.randn() sinh phân phối chuẩn N(0,1) --> N(m,s^2) = m + N(0,1) * s
# Sử dụng hàm np.histogram với density=True để thống kê số lần xuất hiện với số bins = 300 (do dãy số thực)
# Khởi tạo tham số phân phối chuẩn
mu = -1.0
sigma = 1.0
N = 100000
# Sinh ngẫu nhiên N phần tử theo phân phối chuẩn N (mu, sigma^2)
x = mu + sigma * np.random.randn(N)
# Thống kê tần số xuất hiện với histogram(bins = 300), chuẩn hóa mật độ
cnt, val = np.histogram(x, bins=300, density=True)
val = val[:-1]

print( "c. Sinh day so theo phan phoi chuan")
print(f"+ x: len = {len(x)}, min [{np.min(x)}], max [{np.max(x)}]")
print(f"+ Mat so xac suat cua day so x")
plt.figure(figsize=(6,6))      # kich thuoc ban ve
plt.xlim(np.min(val), np.max(val))
plt.scatter(val, cnt)
plt.show()
```

#### 4. Giải hệ phương trình tuyến tính
Giải hệ phương trình tuyến tính sau:
$$
\left\{\begin{matrix}
        4x_1  + 3x_2 - 5x_3 = 2 \\
        -2x_1 - 4x_2 + 5x_3 = 5 \\
        8x_1 + 8x_2 = -3 \\
\end{matrix}\right.
$$

```python
""" CÁC BẠN LÀM BÀI Ở ĐÂY """

# Khai báo ma trận hệ số A và vector kết quả y
A = np.array([
    [4, 3, -5],
    [-2, -4, 5],
    [8, 8, 0]
], dtype=float)

y = np.array([2, 5, -3], dtype=float)

# Giải hệ phương trình Ax = y
x = np.linalg.solve(A, y)

# Kiểm tra nghiệm bằng cách nhân lại: yy = A @ x
yy = A @ x

# Tính sai số chuẩn Euclid giữa y và yy
diff = np.linalg.norm(y - yy)


print(f"+ Ma tran A: \n{A}")
print(f"+ Vector y: {y}")
print(f"+ Nghiem x: {x}")
print(f"+ Kiem tra: yy = Ax = {yy} ==> ||yy -y ||_2 = {diff: .2f}")
```

### Bài 3. Tính gần đúng số pi dùng phương pháp Monte Carlo

Vẽ một đường tròn $C$ có bán kính $r$ nội tiếp một hình vuông $C$, và đặt ngẫu nhiên $N$ chấm lên hình vuông. Tỉ lệ các chấm nằm trong hình tròn trên tổng số chấm xấp xỉ bằng diện tích của hình tròn chia cho hình vuông như sau:
$$
\frac{N_C}{N} \approx  \frac{\mathbb{S}_C}{\mathbb{S}_N}
$$
nơi mà $N_C$ là các chấm trong hình tròn, $N$ là tổng số chấm, $\mathbb{S}_C$ và $\mathbb{S}_N$ lần lượt là diện tích của hình tròn và hình vuông.
$$ \mathbb{S}_C = \pi * r^2 $$
$$ \mathbb{S}_N = (2*r)^2 = 4r^2 $$
Do đó, công thức trên trở thành:
$$
\frac{N_C}{N} \approx \frac{\pi * r^2}{4r^2} = \frac{\pi}{4}
$$
Suy ra, ta có giá trị số $\pi$ được tính như sau:
$$
\pi \approx \frac{4N_C}{N}
$$

Các bạn hãy lập trình tính sấp xỉ số $\pi$ dùng phương pháp Monte Carlo. Cho biết các sai số với $N = 100$, $N = 10000$, và $N = 1000000$. Vẽ đồ thị minh họa cho chương trình.

**Hướng dẫn**: Viết chương trình phát sinh $N$ điểm $(x_i, y_i)$ với $x_i, y_i \in [-1, -1]$. $(x_i, y_i)$ thuộc hình tròn tâm $(0, 0)$ bán kính 1 nội tiếp hình vuông có tâm tại $(0, 0)$ độ dài cạnh 2 khi và chỉ khi $x_i^2 + y_i^2 <= 1$.

```python
import matplotlib.pyplot as plt # khai báo thư viện vẽ pyplot
import math

circle = plt.Circle((0, 0), 1, color='g') # tạo đối tượng vòng tròn

fig, ax = plt.subplots(figsize=(6, 6)) # lấy figure và vùng vẽ

plt.plot(0, 0, 'o', color=(0.9, 0.9, 1.0), alpha=0.8) # vẽ điểm tại tâm (0,0)
ax.add_patch(circle) # thêm vòng tròn vào vùng vẽ

x = 0.75; y = math.sqrt(1 - (x ** 2)) # (x = 0.75, y = căn bậc 2 của 1 - x^2)
plt.arrow(0, 0, x, y)

plt.xlim(-1, 1) # giới hạn trục x
plt.ylim(-1, 1) # giới hạn trục y

plt.savefig('pi.png') # lưu đồ thị xuống tập tin
plt.show()            # hiển thị đồ thị
```

```python
import numpy as np

def calc_pi_monte_carlo(n = 100):
    pi = 0

    # gợi ý: sử dụng np.random.rand(s1, s2, ...) --> sinh ngẫu nhiên các số trong nửa đoạn [0.0, 1.0)
    """ CÁC BẠN LÀM BÀI Ở ĐÂY """
    # ...
    # Sinh ngẫu nhiên n điểm (x, y) trong hình vuông [-1, 1] × [-1, 1]
    x = np.random.rand(n) * 2 - 1  # từ [0,1) → [-1,1)
    y = np.random.rand(n) * 2 - 1

    # Kiểm tra điều kiện nằm trong hình tròn: x^2 + y^2 ≤ 1
    inside_circle = x**2 + y**2 <= 1

    # Đếm số điểm nằm trong hình tròn
    n_inside = np.sum(inside_circle)

    # Tính gần đúng số pi
    pi = 4 * n_inside / n


    return pi
# calc_pi_monte_carlo
```

```python
import math

print("epsilon(n=100): ", calc_pi_monte_carlo(n = 100) - math.pi)
print("epsilon(n=100): ", calc_pi_monte_carlo(n = 100) - math.pi)
print("epsilon(n=10000): ", calc_pi_monte_carlo(n = 10000) - math.pi)
print("epsilon(n=10000): ", calc_pi_monte_carlo(n = 10000) - math.pi)
print("epsilon(n=1000000): ", calc_pi_monte_carlo(n = 1000000) - math.pi)
print("epsilon(n=1000000): ", calc_pi_monte_carlo(n = 1000000) - math.pi)
```
