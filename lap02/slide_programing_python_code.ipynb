```python
#class Vehicle
class Vehicle:
    name = ""
    kind = "car"
    color = ""
    value = 200.00
    def description(self):
        desc_str = "%s is a %s %s worth $%.2f." % (self.name, self.color, self.kind, self.value)
        return desc_str

car1 = Vehicle()
car1.name = "Ford"
car1.color = "black"
car1.kind = "SUV"
car1.value = 15000.00

car2 = Vehicle()
car2.name = "Toyota"
car2.color = "white"
car2.kind = "Sedan"
car2.value = 85000.00
print(car1.name)
print(car1.description())
print(car2.name)
print(car2.description())

```
