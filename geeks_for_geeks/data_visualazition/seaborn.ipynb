# Trực quan hóa dữ liệu với Seaborn - Python

## I. Giới thiệu

- Seaborn là thư viện phổ biến để tạo các hình ảnh trực quan thống kê.
- Xây dựng trên Matplotlib tích hợp Pandas.
- Biểu dồ đường, nhiệt, violin

## II. Tạo biểu đồ với Seaborn

### 1. Biểu đồ đường(Line Plot)

- Thể hiện mối quan hệ hai hay nhiều biến số theo thời giann bằng các đường khác nhau

- Cú Pháp : sns.lineplot(x = none, y = none, data = none )
- Các thông số : 
+ x, y:  Biến dạng số(array, list, col from DataFrame)
+ data : DataFrame chứa dữ liệu

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {'Name' : ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'], 'Age': [21, 23, 20, 24]}
df = pd.DataFrame(data)

plt.plot(df.index, df['Age'])
plt.xlabel('Index')
plt.ylabel('Age')
plt.title('Age Line Plot')
plt.show()
```

### 2. Biểu đồ phân tán(Scatter Plot)

- Sử dụng để trực quan hóa quan hệ giữa 2 biến số
- Giúp xác định mối tương quan hoặc mô hình (có thể có đồ thị 2 chiều)

- Syntax : sns.scatterplot(x = none, y = none, data = none )
- Parameters : x, y : biến dữ liệu đầu vào phải là só.
data (optioanal): bộ dữ liệu chứa các biến
- Trả về một đối tượng Axes có biểu đồ phân tán

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {'Name': ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'], 'Age': [21, 23, 20, 24]}
df = pd.DataFrame(data)

sns.scatterplot(x=df.index, y='Age', data=df)
plt.show()
```

### 3. Biểu đồ hộp(Box plot)

- Biểu diễn trực quan các nhóm dữ liệu số với các phần tư của chúng so với dữ liệu liên tục/phân loại.

- Nó bao gồm 5 số liệu thống kê chính: Tối thiểu, Tứ phân vị thứ nhất hoặc 25%, Trung vị (Tứ phân vị thứ hai) hoặc 50%, Tứ phân vị thứ ba hoặc 75% và Tối đa

- Syntax : sns.boxplot(x= none, y = none, hue = none, data = none)
- Parameters: x, y, hue: Các biến để biểu diễn dữ liệu dạng dài.
dữ liệu: Tập dữ liệu để vẽ đồ thị. Nếu x và y không có, dữ liệu được coi là dữ liệu dạng rộng.
- Returns: An Axes object with the box plot.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {'Name': ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'], 'Age': [21, 23, 20, 45]} 
df = pd.DataFrame(data)

sns.boxplot(y='Age', data=df)
plt.show()
```

### 4. Violin Plot

- Hiển thị nhiều dữ liệu định lượng trên một hoặc nhiều biến phân loại sao cho có thể so sánh được các phân phối.

- Syntax : sns.violinplot(x=none, y=none, hue=none, data=Không có)

- Paramaters : x, y, hue: Đầu vào để vẽ dữ liệu dạng dài. 
- data: Bộ dữ liệu để vẽ đồ thị. 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {'Name': ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'], 'Age': [21, 23, 20, 24]}
df = pd.DataFrame(data)

sns.violinplot(y='Age', data=df)
plt.show()
```

### 5. Swarm plot 

- Hiển thị các điểm dữ liệu riêng lẻ không chồng chéo dọc theo trục phân loại, cung cấp góc nhìn rõ ràng về mật độ phân phối.

Syntax : sns.swarmplot(x=Không có, y=Không có, hue=Không có, data=Không có)

- Parameters: 
- x, y, hue: Đầu vào để vẽ dữ liệu dạng dài. 
- data: Bộ dữ liệu để vẽ đồ thị. 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {'Name': ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'], 'Age': [21, 23, 20, 24]}
df = pd.DataFrame(data)

sns.swarmplot(x=df.index, y='Age', data=df)
plt.show()
```

### 6. Biểu đồ thanh (Barplot)

- Biểu thị ước tính xu hướng trung tâm của một biến số với chiều cao của mỗi hình chữ nhật và cung cấp một số dấu hiệu về sự không chắc chắn xung quanh ước tính đó bằng cách sử dụng các thanh lỗi. 

Syntax:

sns.barplot(x=Không có, y=Không có, hue=Không có, data=Không có)

- Parameters:

- x, y: Biến hoặc tên cột cho dữ liệu dạng dài.
- hue : (tùy chọn) Cột để mã hóa màu.
- data: (tùy chọn) Bộ dữ liệu để vẽ biểu đồ.
- Trả về: Đối tượng trục có biểu đồ thanh.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {'Name': ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'], 'Age': [21, 23, 20, 24]}
df = pd.DataFrame(data)

sns.barplot(x='Name', y='Age', data=df)
plt.show()
```
